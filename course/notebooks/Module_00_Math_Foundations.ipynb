## 0.1 Linear Algebra: Vectors

```python
# Vectors: 1D arrays representing points or directions
v1 = np.array([3, 4])
v2 = np.array([1, 2])

print("Vector Operations:")
print(f"v1 = {v1}")
print(f"v2 = {v2}")
print()
print(f"Addition: v1 + v2 = {v1 + v2}")
print(f"Scalar multiplication: 2 * v1 = {2 * v1}")
print(f"Element-wise: v1 * v2 = {v1 * v2}")
```

```python
# Vector norms (length/magnitude)
def l1_norm(v):
    """L1 norm: sum of absolute values."""
    return np.sum(np.abs(v))

def l2_norm(v):
    """L2 norm: Euclidean distance from origin."""
    return np.sqrt(np.sum(v ** 2))

def linf_norm(v):
    """L-infinity norm: maximum absolute value."""
    return np.max(np.abs(v))

print("Vector Norms:")
print(f"v1 = {v1}")
print(f"||v1||_1 (L1) = {l1_norm(v1)}")
print(f"||v1||_2 (L2) = {l2_norm(v1)}")
print(f"||v1||_inf (L-inf) = {linf_norm(v1)}")
print(f"\nNumPy: np.linalg.norm(v1) = {np.linalg.norm(v1)}")
```

```python
# Dot product and cosine similarity
def dot_product(a, b):
    """Dot product: sum of element-wise products."""
    return np.sum(a * b)

def cosine_similarity(a, b):
    """Cosine of angle between vectors."""
    return dot_product(a, b) / (l2_norm(a) * l2_norm(b))

v3 = np.array([1, 0])
v4 = np.array([0, 1])
v5 = np.array([1, 1])

print("Dot Product and Cosine Similarity:")
print(f"v3 = {v3}, v4 = {v4}, v5 = {v5}")
print()
print(f"v3 . v4 = {dot_product(v3, v4)} (orthogonal)")
print(f"v3 . v5 = {dot_product(v3, v5)}")
print(f"cos(v3, v5) = {cosine_similarity(v3, v5):.4f}")
print(f"Angle = {np.degrees(np.arccos(cosine_similarity(v3, v5))):.1f} degrees")
```

```python
# Visualize vectors
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Vector addition
ax = axes[0]
ax.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v1')
ax.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='red', label='v2')
ax.quiver(0, 0, (v1+v2)[0], (v1+v2)[1], angles='xy', scale_units='xy', scale=1, color='green', label='v1+v2')
ax.quiver(v1[0], v1[1], v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='red', alpha=0.5, linestyle='dashed')
ax.set_xlim(-1, 8)
ax.set_ylim(-1, 8)
ax.grid(True)
ax.legend()
ax.set_title('Vector Addition')
ax.set_aspect('equal')

# Cosine similarity
ax = axes[1]
ax.quiver(0, 0, v3[0], v3[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v3')
ax.quiver(0, 0, v4[0], v4[1], angles='xy', scale_units='xy', scale=1, color='red', label='v4 (orthogonal)')
ax.quiver(0, 0, v5[0], v5[1], angles='xy', scale_units='xy', scale=1, color='green', label='v5')
ax.set_xlim(-0.5, 1.5)
ax.set_ylim(-0.5, 1.5)
ax.grid(True)
ax.legend()
ax.set_title('Cosine Similarity\n(angle between vectors)')
ax.set_aspect('equal')

plt.tight_layout()
plt.show()
```

## 0.2 Linear Algebra: Matrices

```python
# Matrix creation and basic operations
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[1, 4],
              [2, 5],
              [3, 6]])

print(f"Matrix A (2x3):\n{A}")
print(f"\nMatrix B (3x2):\n{B}")
print(f"\nA shape: {A.shape}")
print(f"B shape: {B.shape}")
```

```python
# Matrix multiplication
C = A @ B  # or np.matmul(A, B) or np.dot(A, B)

print("Matrix Multiplication: C = A @ B")
print(f"A shape: {A.shape}, B shape: {B.shape}")
print(f"C shape: {C.shape}")
print(f"\nC = \n{C}")
print("\nComputation:")
print(f"C[0,0] = A[0,:] . B[:,0] = {A[0]} . {B[:,0]} = {C[0,0]}")
```

```python
# Matrix-vector multiplication (linear transformation)
W = np.array([[1, 0],
              [0, 2]])  # Scaling matrix

x = np.array([1, 1])
y = W @ x

print("Matrix-Vector Multiplication (Linear Transform):")
print(f"W = \n{W}")
print(f"x = {x}")
print(f"y = W @ x = {y}")
print("\nThis matrix scales x by 1 and y by 2.")
```

```python
# Visualize linear transformations
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Create unit circle points
theta = np.linspace(0, 2*np.pi, 100)
circle = np.array([np.cos(theta), np.sin(theta)])

transforms = [
    (np.array([[1, 0], [0, 1]]), "Identity"),
    (np.array([[2, 0], [0, 0.5]]), "Scale (2, 0.5)"),
    (np.array([[np.cos(np.pi/4), -np.sin(np.pi/4)], 
               [np.sin(np.pi/4), np.cos(np.pi/4)]]), "Rotate 45°")
]

for ax, (M, title) in zip(axes, transforms):
    transformed = M @ circle
    ax.plot(circle[0], circle[1], 'b-', label='Original', alpha=0.5)
    ax.plot(transformed[0], transformed[1], 'r-', label='Transformed')
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.legend()
    ax.set_title(f"{title}\nM = {M.tolist()}")

plt.suptitle('Linear Transformations', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Eigenvalues and eigenvectors
M = np.array([[3, 1],
              [1, 3]])

eigenvalues, eigenvectors = np.linalg.eig(M)

print("Eigenvalues and Eigenvectors:")
print(f"M = \n{M}")
print(f"\nEigenvalues: {eigenvalues}")
print(f"Eigenvectors:\n{eigenvectors}")

# Verify: M @ v = lambda * v
for i in range(2):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    print(f"\nVerify: M @ v{i+1} = {M @ v}, lambda * v{i+1} = {lam * v}")
```

## 0.3 Calculus: Derivatives and Gradients

```python
# Derivative: rate of change
def numerical_derivative(f, x, h=1e-5):
    """Compute derivative using finite differences."""
    return (f(x + h) - f(x - h)) / (2 * h)

# Example function: f(x) = x^2
f = lambda x: x ** 2
f_derivative = lambda x: 2 * x  # Analytical derivative

x = 3.0
print("Derivatives:")
print(f"f(x) = x^2 at x = {x}")
print(f"Analytical: f'(x) = 2x = {f_derivative(x)}")
print(f"Numerical:  f'(x) ≈ {numerical_derivative(f, x):.6f}")
```

```python
# Visualize derivative
x_range = np.linspace(-3, 3, 100)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Function and tangent line
x0 = 1.5
slope = f_derivative(x0)
tangent = slope * (x_range - x0) + f(x0)

axes[0].plot(x_range, f(x_range), 'b-', label='f(x) = x²')
axes[0].plot(x_range, tangent, 'r--', label=f'Tangent at x={x0}')
axes[0].scatter([x0], [f(x0)], color='red', s=100, zorder=5)
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
axes[0].set_title('Function and Tangent Line')

# Function and its derivative
axes[1].plot(x_range, f(x_range), 'b-', label='f(x) = x²')
axes[1].plot(x_range, f_derivative(x_range), 'r-', label="f'(x) = 2x")
axes[1].axhline(y=0, color='k', linewidth=0.5)
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].legend()
axes[1].grid(True, alpha=0.3)
axes[1].set_title('Function and Derivative')

plt.tight_layout()
plt.show()
```

```python
# Chain rule: derivative of composed functions
# If h(x) = f(g(x)), then h'(x) = f'(g(x)) * g'(x)

# Example: h(x) = (x^2 + 1)^3
# g(x) = x^2 + 1, f(u) = u^3
# g'(x) = 2x, f'(u) = 3u^2
# h'(x) = 3(x^2 + 1)^2 * 2x = 6x(x^2 + 1)^2

g = lambda x: x**2 + 1
f_outer = lambda u: u**3
h = lambda x: f_outer(g(x))  # Composed function

# Analytical derivative using chain rule
h_derivative = lambda x: 6 * x * (x**2 + 1)**2

x = 2.0
print("Chain Rule:")
print(f"h(x) = (x² + 1)³")
print(f"h'(x) = 3(x² + 1)² · 2x = 6x(x² + 1)²")
print(f"\nAt x = {x}:")
print(f"  Analytical: h'({x}) = {h_derivative(x)}")
print(f"  Numerical:  h'({x}) ≈ {numerical_derivative(h, x):.6f}")
```

```python
# Gradient: derivative for multivariable functions
# f(x, y) = x^2 + y^2
# grad f = [∂f/∂x, ∂f/∂y] = [2x, 2y]

def f_2d(xy):
    x, y = xy
    return x**2 + y**2

def gradient_f_2d(xy):
    x, y = xy
    return np.array([2*x, 2*y])

def numerical_gradient(f, xy, h=1e-5):
    """Compute gradient numerically."""
    grad = np.zeros(len(xy))
    for i in range(len(xy)):
        xy_plus = xy.copy()
        xy_minus = xy.copy()
        xy_plus[i] += h
        xy_minus[i] -= h
        grad[i] = (f(xy_plus) - f(xy_minus)) / (2 * h)
    return grad

point = np.array([1.0, 2.0])
print("Gradient (Multivariable Derivative):")
print(f"f(x, y) = x² + y²")
print(f"∇f = [∂f/∂x, ∂f/∂y] = [2x, 2y]")
print(f"\nAt point {point}:")
print(f"  Analytical: ∇f = {gradient_f_2d(point)}")
print(f"  Numerical:  ∇f ≈ {numerical_gradient(f_2d, point)}")
```

```python
# Visualize gradient
fig, ax = plt.subplots(figsize=(8, 8))

# Create grid
x_range = np.linspace(-3, 3, 30)
y_range = np.linspace(-3, 3, 30)
X, Y = np.meshgrid(x_range, y_range)
Z = X**2 + Y**2

# Contour plot
cs = ax.contour(X, Y, Z, levels=10, cmap='viridis')
ax.clabel(cs, inline=True, fontsize=8)

# Gradient vectors (pointing towards steepest ascent)
x_sparse = np.linspace(-2.5, 2.5, 8)
y_sparse = np.linspace(-2.5, 2.5, 8)
Xs, Ys = np.meshgrid(x_sparse, y_sparse)
U = 2 * Xs  # ∂f/∂x
V = 2 * Ys  # ∂f/∂y

ax.quiver(Xs, Ys, U, V, color='red', alpha=0.7)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('f(x,y) = x² + y²\nContours with Gradient Vectors (red)')
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)

plt.show()
```

## 0.4 Probability Basics

```python
# Probability distributions
def gaussian_pdf(x, mu=0, sigma=1):
    """Gaussian (Normal) probability density function."""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

x = np.linspace(-5, 5, 200)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Different means
for mu in [-1, 0, 1]:
    axes[0].plot(x, gaussian_pdf(x, mu=mu), label=f'μ={mu}, σ=1')
axes[0].set_xlabel('x')
axes[0].set_ylabel('p(x)')
axes[0].legend()
axes[0].set_title('Gaussian: Varying Mean')
axes[0].grid(True, alpha=0.3)

# Different standard deviations
for sigma in [0.5, 1, 2]:
    axes[1].plot(x, gaussian_pdf(x, sigma=sigma), label=f'μ=0, σ={sigma}')
axes[1].set_xlabel('x')
axes[1].set_ylabel('p(x)')
axes[1].legend()
axes[1].set_title('Gaussian: Varying Std Dev')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Mean, variance, standard deviation
samples = np.random.randn(1000) * 2 + 3  # Normal(3, 4)

print("Sample Statistics:")
print(f"  Mean (E[X]): {np.mean(samples):.3f} (expected: 3)")
print(f"  Variance (Var[X]): {np.var(samples):.3f} (expected: 4)")
print(f"  Std Dev (σ): {np.std(samples):.3f} (expected: 2)")
```

```python
# Bayes' theorem
# P(A|B) = P(B|A) * P(A) / P(B)

# Example: Medical test
# Disease prevalence: P(D) = 0.01 (1% have disease)
# Test sensitivity: P(+|D) = 0.99 (99% true positive)
# Test specificity: P(-|¬D) = 0.95 (95% true negative)

P_D = 0.01  # Prior: probability of disease
P_pos_given_D = 0.99  # Likelihood: P(+|D)
P_neg_given_not_D = 0.95  # P(-|¬D)
P_pos_given_not_D = 1 - P_neg_given_not_D  # P(+|¬D) = 0.05

# P(+) = P(+|D)P(D) + P(+|¬D)P(¬D)
P_pos = P_pos_given_D * P_D + P_pos_given_not_D * (1 - P_D)

# Bayes: P(D|+) = P(+|D)P(D) / P(+)
P_D_given_pos = P_pos_given_D * P_D / P_pos

print("Bayes' Theorem Example: Medical Test")
print("=" * 50)
print(f"Prior P(Disease) = {P_D}")
print(f"P(Positive | Disease) = {P_pos_given_D}")
print(f"P(Positive | No Disease) = {P_pos_given_not_D}")
print()
print(f"P(Positive) = {P_pos:.4f}")
print(f"P(Disease | Positive) = {P_D_given_pos:.4f}")
print()
print(f"Even with 99% sensitive test, a positive result only")
print(f"means ~{P_D_given_pos*100:.1f}% chance of disease!")
```

## 0.5 Information Theory

```python
# Entropy: measure of uncertainty
def entropy(probs):
    """
    Shannon entropy: H(p) = -sum(p * log(p))
    
    Higher entropy = more uncertainty.
    """
    probs = np.array(probs)
    probs = probs[probs > 0]  # Avoid log(0)
    return -np.sum(probs * np.log2(probs))

# Examples
distributions = [
    ([0.5, 0.5], "Fair coin"),
    ([0.9, 0.1], "Biased coin (90%)"),
    ([1.0, 0.0], "Certain outcome"),
    ([0.25, 0.25, 0.25, 0.25], "4-sided fair die"),
]

print("Entropy Examples:")
print("-" * 50)
for probs, name in distributions:
    H = entropy(probs)
    print(f"{name}: H = {H:.4f} bits")
```

```python
# Cross-entropy: measures how well q approximates p
def cross_entropy(p, q):
    """
    Cross-entropy: H(p, q) = -sum(p * log(q))
    
    Used as loss function in classification.
    """
    p = np.array(p)
    q = np.array(q)
    q = np.clip(q, 1e-10, 1)  # Avoid log(0)
    return -np.sum(p * np.log(q))

# True distribution
p_true = [1, 0, 0]  # One-hot: class 0 is correct

# Different predictions
predictions = [
    ([0.99, 0.005, 0.005], "Confident correct"),
    ([0.7, 0.2, 0.1], "Somewhat correct"),
    ([0.33, 0.33, 0.34], "Uncertain"),
    ([0.1, 0.8, 0.1], "Wrong"),
]

print("Cross-Entropy (Classification Loss):")
print(f"True class: 0 (one-hot: {p_true})")
print("-" * 50)
for q_pred, name in predictions:
    ce = cross_entropy(p_true, q_pred)
    print(f"{name}: q={q_pred} -> CE={ce:.4f}")
```

```python
# KL Divergence: measures difference between distributions
def kl_divergence(p, q):
    """
    KL Divergence: KL(p || q) = sum(p * log(p/q))
    
    KL(p || q) = H(p, q) - H(p)
    
    KL >= 0, and KL = 0 iff p = q.
    Note: KL is not symmetric!
    """
    p = np.array(p)
    q = np.array(q)
    mask = p > 0
    return np.sum(p[mask] * np.log(p[mask] / (q[mask] + 1e-10)))

# Compare distributions
p1 = [0.5, 0.5]
p2 = [0.6, 0.4]
p3 = [0.9, 0.1]

print("KL Divergence:")
print("-" * 50)
print(f"KL([0.5, 0.5] || [0.5, 0.5]) = {kl_divergence(p1, p1):.4f}")
print(f"KL([0.5, 0.5] || [0.6, 0.4]) = {kl_divergence(p1, p2):.4f}")
print(f"KL([0.6, 0.4] || [0.5, 0.5]) = {kl_divergence(p2, p1):.4f} (asymmetric!)")
print(f"KL([0.5, 0.5] || [0.9, 0.1]) = {kl_divergence(p1, p3):.4f} (bigger diff)")
```

```python
# Visualize KL divergence
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Binary distribution KL as function of q
p_fixed = 0.3
q_range = np.linspace(0.01, 0.99, 100)
kl_values = [kl_divergence([p_fixed, 1-p_fixed], [q, 1-q]) for q in q_range]

axes[0].plot(q_range, kl_values)
axes[0].axvline(x=p_fixed, color='r', linestyle='--', label=f'p={p_fixed}')
axes[0].set_xlabel('q')
axes[0].set_ylabel('KL(p || q)')
axes[0].set_title(f'KL Divergence: p=[{p_fixed}, {1-p_fixed}]')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Compare P and Q distributions
x = [0, 1]
width = 0.35
axes[1].bar([i - width/2 for i in x], [0.3, 0.7], width, label='P=[0.3, 0.7]', color='blue', alpha=0.7)
axes[1].bar([i + width/2 for i in x], [0.5, 0.5], width, label='Q=[0.5, 0.5]', color='red', alpha=0.7)
axes[1].set_xticks(x)
axes[1].set_xticklabels(['Class 0', 'Class 1'])
axes[1].set_ylabel('Probability')
axes[1].legend()
kl_pq = kl_divergence([0.3, 0.7], [0.5, 0.5])
kl_qp = kl_divergence([0.5, 0.5], [0.3, 0.7])
axes[1].set_title(f'KL(P||Q)={kl_pq:.3f}, KL(Q||P)={kl_qp:.3f}')

plt.tight_layout()
plt.show()
```

## 0.6 Softmax Function

```python
def softmax(x):
    """
    Softmax: converts logits to probabilities.
    
    softmax(x)_i = exp(x_i) / sum(exp(x_j))
    
    Properties:
    - Output sums to 1
    - All outputs in (0, 1)
    - Preserves order
    """
    # Subtract max for numerical stability
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

# Example
logits = np.array([2.0, 1.0, 0.1])
probs = softmax(logits)

print("Softmax:")
print(f"Logits: {logits}")
print(f"Probabilities: {probs}")
print(f"Sum: {probs.sum():.6f}")
```

```python
# Temperature scaling
def softmax_with_temp(x, temperature=1.0):
    """Softmax with temperature."""
    x = x / temperature
    return softmax(x)

logits = np.array([2.0, 1.0, 0.5])
temperatures = [0.5, 1.0, 2.0, 5.0]

fig, ax = plt.subplots(figsize=(10, 6))

x_pos = np.arange(3)
width = 0.2

for i, temp in enumerate(temperatures):
    probs = softmax_with_temp(logits, temp)
    ax.bar(x_pos + i * width, probs, width, label=f'T={temp}')

ax.set_xticks(x_pos + width * 1.5)
ax.set_xticklabels(['Class 0', 'Class 1', 'Class 2'])
ax.set_ylabel('Probability')
ax.set_title(f'Softmax Temperature Scaling\nLogits: {logits}')
ax.legend()
ax.grid(True, alpha=0.3, axis='y')

plt.show()

print("\nTemperature effects:")
print("  T < 1: Sharper (more confident)")
print("  T = 1: Standard softmax")
print("  T > 1: Flatter (more uniform)")
print("  T → 0: Argmax (one-hot)")
print("  T → ∞: Uniform distribution")
```
